### 📖 Conceptual Foundations

#### Agglomerative Hierarchical Clustering

A **bottom-up** approach where each observation starts in its own individual cluster. At each step, the two closest clusters are iteratively merged according to a **linkage criterion** until all points belong to a single root cluster, visualized as a **Dendrogram**.


#### Linkage Criteria for Hierarchical Merging

The distance between two clusters $A$ and $B$ is defined by the linkage method:

- **Ward's Linkage (Default in Scikit-learn):** Minimizes the total within-cluster variance increase resulting from merging clusters. Produces compact, spherical clusters similar to K-Means.
- **Complete Linkage (Maximum):** Distance is the **maximum distance** between any point in $A$ and any point in $B$: $\max \{d(a, b) : a \in A, b \in B\}$. Avoids long chaining; produces compact clusters.
- **Single Linkage (Minimum):** Distance is the **minimum distance** between any point in $A$ and any point in $B$: $\min \{d(a, b) : a \in A, b \in B\}$. Can trace non-elliptical shapes, but suffers from chaining artifacts.
- **Average Linkage:** Average distance between all pairs: $\frac{1}{|A||B|} \sum_{a \in A} \sum_{b \in B} d(a, b)$.


#### DBSCAN: Density-Based Spatial Clustering of Applications with Noise

Pioneered by Ester et al. (1996), DBSCAN groups points based on spatial **density reachability** rather than distance to centroids. It requires two hyperparameters:
- **$\epsilon$ (eps):** Radius of the neighborhood around a point.
- **`min_samples`:** Minimum number of points required within $\epsilon$-radius to form a dense region.

| Point Classification | Condition | Role in Algorithm |
|---|---|---|
| **Core Point** | Has $\ge \text{min\_samples}$ within $\epsilon$-radius | Forms the dense foundation of a cluster |
| **Border Point** | Has $< \text{min\_samples}$ within $\epsilon$, but within $\epsilon$ of a Core Point | Belongs to the cluster on the perimeter |
| **Noise Point (Outlier)** | Neither a Core Point nor a Border Point | Labelled as **-1** (pure anomaly / outlier) |


#### Why DBSCAN Beats K-Means on Complex Geometry

DBSCAN **does not require specifying $K$ in advance**, can find **arbitrary non-linear shapes** (concentric rings, smiling faces, road networks), and **naturally isolates outliers as noise (-1)** instead of forcing them into clusters!


#### Core Algorithm Selection

Use **Agglomerative Hierarchical** when taxonomy or dendrogram hierarchies are needed (e.g. biological phylogenetics, document taxonomies). Use **DBSCAN / HDBSCAN** when data has non-convex geometries and noisy spatial outliers.



```python
# %matplotlib inline
import numpy as np
from sklearn.datasets import make_circles
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler

# Generate concentric circles (non-linear manifold)
X, _ = make_circles(n_samples=300, factor=0.4, noise=0.05, random_state=42)
X_scaled = StandardScaler().fit_transform(X)

# 1. K-Means (Fails on non-convex data)
km = KMeans(n_clusters=2, random_state=42, n_init='auto')
km_labels = km.fit_predict(X_scaled)

# 2. DBSCAN (Succeeds via density reachability)
db = DBSCAN(eps=0.3, min_samples=5)
db_labels = db.fit_predict(X_scaled)

n_clusters_db = len(set(db_labels)) - (1 if -1 in db_labels else 0)
n_noise = list(db_labels).count(-1)

print(f'K-Means forced 2 clusters with linear boundary split.')
print(f'DBSCAN automatically found {n_clusters_db} clusters and {n_noise} noise points!')
print(f'DBSCAN unique labels: {set(db_labels)}')
```
