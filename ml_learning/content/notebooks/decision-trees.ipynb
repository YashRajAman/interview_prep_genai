### 📖 Conceptual Foundations

#### Core Algorithm Mechanism (CART)

A Decision Tree recursively partitions feature space into axis-aligned rectangular regions. Predictions are made by passing a query down the tree until reaching a terminal leaf node:
- **Classification:** Predicts the **majority class** (mode) of training samples in the leaf.
- **Regression:** Predicts the **mean target value** of samples in the leaf.


#### Mathematical Splitting Criteria for Classification

At each node, the algorithm greedily searches across all features $j$ and split thresholds $t$ to maximize purity:

1. **Gini Impurity (CART Default):** Measures the probability of misclassifying a randomly chosen element:
$$I_G(p) = 1 - \sum_{k=1}^K p_k^2$$
- Pure node (all one class): $I_G = 0$.
- Maximum impurity (equal split in binary): $I_G = 0.5$.

2. **Entropy & Information Gain (ID3 / C4.5):** Measures information disorder in bits:
$$H(p) = -\sum_{k=1}^K p_k \log_2(p_k)$$
$$\text{Information Gain} = H(D) - \left( \frac{|D_L|}{|D|} H(D_L) + \frac{|D_R|}{|D|} H(D_R) \right)$$


#### Gini Impurity vs. Entropy

| Metric | Mathematical Formula | Computation Speed | Behavior |
|---|---|---|---|
| **Gini Impurity** | $1 - \sum p_k^2$ | Faster (no logarithms) | CART standard; tends to isolate dominant classes |
| **Entropy** | $-\sum p_k \log_2(p_k)$ | Slightly slower (log calculations) | Information theoretic; produces slightly more balanced trees |


#### Splitting Criterion for Regression

For continuous regression, trees split by minimizing **Variance Reduction** (equivalent to Mean Squared Error):

$$\text{MSE}_{\text{node}} = \frac{1}{n} \sum_{i=1}^n (y_i - \bar{y}_{\text{node}})^2$$
The split $(j, t)$ is chosen to minimize the weighted sum of left and right child MSE.


#### Primary Vulnerability: Overfitting & Instability

Unconstrained decision trees will continue splitting until every leaf contains a single sample, achieving 100% training accuracy but memorizing noise! **Remedy:** Pre-pruning (`max_depth`, `min_samples_split`, `min_samples_leaf`) or post-pruning via cost-complexity pruning ($\alpha$).


#### Key Architectural Advantage

Decision trees require **zero feature scaling**, naturally handle mixed continuous and categorical data, and provide complete white-box interpretability through visual tree diagrams.



```python
# %matplotlib inline
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.model_selection import train_test_split

# Load iris dataset
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.3, random_state=42)

# 1. Train an interpretable constrained tree (max_depth=3)
clf = DecisionTreeClassifier(max_depth=3, criterion='gini', random_state=42)
clf.fit(X_train, y_train)

print(f'Train Accuracy: {clf.score(X_train, y_train)*100:.2f}%')
print(f'Test Accuracy:  {clf.score(X_test, y_test)*100:.2f}%')

print('\n=== Feature Importances (Gini Importance) ===')
for name, imp in zip(iris.feature_names, clf.feature_importances_):
    print(f'{name:20s}: {imp:.4f}')

print('\n=== Exported Tree Decision Rules ===')
print(export_text(clf, feature_names=list(iris.feature_names)))
```
