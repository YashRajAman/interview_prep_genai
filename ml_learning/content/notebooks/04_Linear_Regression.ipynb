# 📈 Linear Regression & Residuals
Explore linear model fitting and regression diagnostics.


```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

np.random.seed(42)
X = np.linspace(0, 10, 80).reshape(-1, 1)
y = 3.2 * X.squeeze() + 4.5 + np.random.normal(0, 3.0, size=80)

model = LinearRegression().fit(X, y)
y_pred = model.predict(X)

print(f"Learned Slope (w): {model.coef_[0]:.3f} (True: 3.200)")
print(f"Learned Intercept (b): {model.intercept_:.3f} (True: 4.500)")
print(f"R2 Score: {r2_score(y, y_pred):.4f}")
```

```python
plt.figure(figsize=(9, 5))
plt.scatter(X, y, color="#00d4aa", alpha=0.8, label="Data Points")
plt.plot(X, y_pred, color="#ff6b6b", lw=2.5, label="Fitted Line")
plt.title("Linear Regression Line Fit")
plt.legend()
plt.tight_layout()
plt.show()
```
