```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="darkgrid")
print("Libraries loaded successfully!")
```

```python
# Create a synthetic customer dataset
np.random.seed(42)
n = 300

df = pd.DataFrame({
    "customer_id": range(1001, 1001 + n),
    "age": np.random.randint(18, 70, size=n),
    "income": np.random.normal(72000, 24000, size=n).round(-2),
    "device": np.random.choice(["iOS", "Android", "Web"], size=n, p=[0.45, 0.40, 0.15]),
    "monthly_spend": np.random.gamma(shape=3.0, scale=45.0, size=n).round(2),
    "churned": np.random.choice([0, 1], size=n, p=[0.82, 0.18])
})

df.head(10)
```

```python
# Summary statistics
%time df.describe().round(2)
```

```python
# Device aggregation
df.groupby("device")[["income", "monthly_spend", "churned"]].mean().round(3)
```

```python
# Seaborn scatterplot
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df,
    x="income",
    y="monthly_spend",
    hue="device",
    style="churned",
    palette="Set2",
    alpha=0.85
)
plt.title("Customer Spending vs. Income by Device", fontsize=13)
plt.tight_layout()
plt.show()
```
