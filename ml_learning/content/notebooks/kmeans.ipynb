### 📖 Conceptual Foundations

#### Mathematical Objective: Inertia (WCSS)

Given an unlabeled dataset $X = \{\mathbf{x}_1, \dots, \mathbf{x}_n\}$, K-Means partitions the $n$ observations into $K$ disjoint sets $S = \{S_1, S_2, \dots, S_K\}$ to minimize the **Within-Cluster Sum of Squares (Inertia)**:

$$J = \sum_{k=1}^K \sum_{\mathbf{x} \in S_k} \|\mathbf{x} - \boldsymbol{\mu}_k\|^2$$

where $\boldsymbol{\mu}_k = \frac{1}{|S_k|} \sum_{\mathbf{x} \in S_k} \mathbf{x}$ is the arithmetic mean centroid of cluster $S_k$.


#### Lloyd's Alternating Minimization Algorithm

Finding the global minimum of $J$ is NP-hard. Lloyd's heuristic converges iteratively:
1. **Initialization:** Select $K$ initial centroid coordinates $\boldsymbol{\mu}_1, \dots, \boldsymbol{\mu}_K$.
2. **Assignment Step (Expectation):** Assign every observation $\mathbf{x}_i$ to its nearest centroid:
$$c^{(i)} = \arg\min_k \|\mathbf{x}_i - \boldsymbol{\mu}_k\|^2$$
3. **Update Step (Maximization):** Recalculate each centroid as the mean of all points assigned to it:
$$\boldsymbol{\mu}_k = \frac{1}{|S_k|} \sum_{i \in S_k} \mathbf{x}_i$$
4. **Repeat** steps 2 and 3 until centroids cease moving (convergence).


#### K-Means++ Initialization (Arthur & Vassilvitskii, 2007)

Standard random initialization frequently converges to suboptimal, high-inertia local minima.

**K-Means++ Smart Initialization:**
1. Choose the first centroid $\boldsymbol{\mu}_1$ uniformly at random from the dataset.
2. For each remaining point $\mathbf{x}$, compute $D(\mathbf{x})$: the distance to the nearest already-chosen centroid.
3. Choose the next centroid with probability proportional to squared distance: $P(\mathbf{x}) = \frac{D(\mathbf{x})^2}{\sum D(\mathbf{x}')^2}$.
4. Repeat until $K$ centroids are chosen. **Guarantees $O(\log K)$ competitive bound against optimal clustering!**


#### Determining Optimal K: Elbow Method vs. Silhouette Score

- **The Elbow Method:** Plot Inertia vs. $K$. As $K$ increases, inertia monotonically decreases. Look for the 'inflection elbow' where marginal variance reduction sharply flattens.
- **Silhouette Analysis:** Measures how similar point $i$ is to its own cluster ($a(i)$) compared to the nearest neighboring cluster ($b(i)$):
$$s(i) = \frac{b(i) - a(i)}{\max(a(i), b(i)}} \in [-1, +1]$$
  - $s \approx +1$: Sample is well clustered far from neighbors.
  - $s \approx 0$: Sample is on the decision boundary between clusters.
  - $s < 0$: Sample is likely assigned to the wrong cluster.


#### Core Limitations of K-Means

1. Assumes clusters are **spherical (isotropic)** and of roughly equal size.
2. Fails completely on non-convex geometric shapes (e.g. concentric rings or half-moons).
3. Highly sensitive to extreme outliers, which pull centroids away from true cluster cores.


#### Preprocessing Mandate

Because K-Means optimizes squared Euclidean distances, **feature standardization (StandardScaler) is strictly mandatory**. Unscaled features will distort cluster shapes along the largest numerical axis.



```python
# %matplotlib inline
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Generate synthetic data with 4 true clusters
X, _ = make_blobs(n_samples=200, centers=4, n_features=2, cluster_std=0.8, random_state=42)

print('=== Evaluating Cluster Quality Across K = 2 to 6 ===')
for k in range(2, 7):
    km = KMeans(n_clusters=k, init='k-means++', n_init=10, random_state=42)
    labels = km.fit_predict(X)
    sil = silhouette_score(X, labels)
    print(f'K = {k}: Inertia (WCSS) = {km.inertia_:8.2f} | Silhouette Score = {sil:.4f}')
```
