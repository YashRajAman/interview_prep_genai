### 📖 Conceptual Foundations

#### What is EDA in Simple Terms?

Imagine buying a used car. You wouldn't just wire money without looking under the hood, checking the odometer, and inspecting the tires!

**Exploratory Data Analysis (EDA)** is doing the exact same vehicle inspection on your dataset. Before feeding data into an algorithm, data scientists create simple visual charts and summary numbers to **spot patterns, discover mistakes, and see what the data looks like**.


#### The 4 Essential Charts for Beginners

| Chart Type | What It Looks Like | Best Used For | Everyday Example |
| --- | --- | --- | --- |
| Bar Chart | Vertical or horizontal bars of different lengths. | Comparing counts or totals across categories. | Comparing total phone sales by Brand (Apple vs Samsung vs Google). |
| Histogram | Bars touching each other showing how numbers cluster. | Seeing how a single numeric measurement is spread out. | Looking at the age distribution of website visitors (Are they teens or seniors?). |
| Scatter Plot | A cloud of individual dots on an X and Y grid. | Showing the relationship between TWO numbers. | Plotting House Size (X) vs House Price (Y) to see if bigger houses cost more. |
| Line Chart | A continuous line moving from left to right. | Tracking how a number changes over time. | Showing daily stock prices or temperature over 12 months. |


#### What You Are Hunting For During EDA

1. **Obvious Typos:** Someone entered an age of `-5` or `999`.
2. **Missing Blanks:** Whole columns where half the rows are empty.
3. **Surprising Patterns:** Do sales always spike on Friday evenings?
4. **Unbalanced Groups:** 99% of customers are from California and only 1% from other states.



```python
# %matplotlib inline
import matplotlib.pyplot as plt

# 1. Bar Chart: Comparing counts across categories
cities = ['New York', 'London', 'Tokyo', 'Paris']
visitors = [450, 380, 520, 310]

plt.figure(figsize=(7, 3))
plt.subplot(1, 2, 1)
plt.bar(cities, visitors, color=['orange', 'teal', 'coral', 'skyblue'])
plt.title('Visitors by City')
plt.ylabel('Count')

# 2. Scatter Plot: Showing relationship between two numbers
hours = [1, 2, 3, 4, 5, 6, 7, 8]
scores = [52, 58, 65, 68, 76, 83, 89, 95]

plt.subplot(1, 2, 2)
plt.scatter(hours, scores, color='purple', s=60)
plt.title('Study Hours vs Exam Score')
plt.xlabel('Hours'); plt.ylabel('Score')

plt.tight_layout()
plt.show()
```
