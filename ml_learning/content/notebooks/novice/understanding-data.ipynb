### 📖 Conceptual Foundations

#### Data is the Food that Feeds Machine Learning

Machine learning models cannot look out the window or read minds—they only learn from the data we give them! Usually, this data looks like a familiar **spreadsheet** (or CSV file):

- **Rows (Examples):** Each row represents one single person, house, patient, or item.
- **Columns (Features):** Each column represents a specific measurement or clue (like Age, Income, or City).


#### The Two Main Types of Data

| Data Type | What is it? | Everyday Examples | How Computers Treat It |
| --- | --- | --- | --- |
| Numerical (Numbers) | Measurements you can add, subtract, and count. | House size (sq ft), Age in years, Temperature, Bank balance ($). | Computers understand numbers directly. |
| Categorical (Categories) | Labels, names, or groups that describe words or types. | City (New York, Paris, Tokyo), Color (Red, Blue), Yes/No, Pet (Dog, Cat). | Must be converted to numbers (like 0, 1, 2) before feeding to math models. |


#### Missing Values & Outliers in Real Life

- **Missing Values:** What happens when someone leaves a question blank on a survey? You get an empty box (`NaN` or null). We must either drop that row or fill in the blank with a sensible value (like the average).
- **Outliers:** A data point that is bizarrely different from the rest. For example, if a customer's age is entered as `250`, it's almost certainly a typing mistake for `25`!



```python
# %matplotlib inline
import pandas as pd
import numpy as np

# Create a customer dataset with 5 rows and 5 columns
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [28, 35, np.nan, 42, 29], # Charlie's age is blank!
    'City': ['Chicago', 'Dallas', 'Chicago', 'Miami', 'Chicago'],
    'Balance': [1500, 3200, 4800, 95000, 2100], # Diana has high balance
    'Subscribed': [1, 0, 1, 1, 0]
}

df = pd.DataFrame(data)
print('=== Full Customer Dataset ===')
print(df)

print('\n=== Checking for Missing Empty Boxes ===')
print(df.isnull().sum())

print('\n=== Filling Missing Age with the Group Average ===')
avg_age = df['Age'].mean()
df['Age'] = df['Age'].fillna(avg_age)
print(df[['Name', 'Age']])
```
