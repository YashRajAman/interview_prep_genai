### 📖 Conceptual Foundations

#### Bayesian Classification Formulation

Given a feature vector $\mathbf{x} = [x_1, x_2, \dots, x_d]^T$, we want to find the class $y \in \{1, \dots, K\}$ that maximizes the posterior probability $P(y | \mathbf{x})$:

$$P(y | \mathbf{x}) = \frac{P(\mathbf{x} | y) P(y)}{P(\mathbf{x})} \propto P(y) P(x_1, x_2, \dots, x_d | y)$$

Computing the full joint likelihood $P(x_1, \dots, x_d | y)$ requires estimating $2^d$ parameters, which is computationally intractable.


#### The 'Naive' Conditional Independence Assumption

Naive Bayes makes the simplifying assumption that **all features are conditionally independent given the class label $y$**:

$$P(x_1, x_2, \dots, x_d | y) = \prod_{j=1}^d P(x_j | y)$$

The classification decision rule is therefore:
$$\hat{y} = \arg\max_y \left[ P(y) \prod_{j=1}^d P(x_j | y) \right] = \arg\max_y \left[ \ln P(y) + \sum_{j=1}^d \ln P(x_j | y) \right]$$


#### The Three Primary Naive Bayes Variants

| Variant | Feature Modality | Likelihood Model $P(x_j|y)$ | Standard Industry Use-Case |
|---|---|---|---|
| **Gaussian NB** | Continuous numeric features | Gaussian PDF: $\frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(x - \mu)^2}{2\sigma^2}}$ | Sensor readings, biometric measurements |
| **Multinomial NB** | Discrete count data | Multinomially distributed word/token counts | Document classification, bag-of-words text |
| **Bernoulli NB** | Binary presence/absence | Bernoulli trial: $p_j^x (1-p_j)^{1-x}$ | Keyword presence in spam, short message tagging |


#### The Zero Frequency Problem & Laplace Smoothing

If a word appears in the test document that never occurred in training class $y$, $P(\text{word} | y) = 0$. Because the likelihood is a product, **a single zero multiplier wipes out all other evidence and sets the entire probability to zero!**

**Remedy: Laplace (Additive) Smoothing:**
$$P(x_j = w | y) = \frac{N_{yj} + \alpha}{N_y + \alpha |V|}$$
Setting $\alpha = 1.0$ guarantees no word ever receives a probability of zero.


#### Why Naive Bayes Works Surprisingly Well

Even when the conditional independence assumption is mathematically violated in practice, Naive Bayes often produces surprisingly accurate **classification boundaries**, because finding the maximum posterior class requires only the correct ranking, not perfectly calibrated probabilities.



```python
# %matplotlib inline
from sklearn.datasets import load_iris
from sklearn.naive_bayes import GaussianNB, MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split

# 1. Gaussian Naive Bayes on Continuous Data
iris = load_iris()
X_tr, X_te, y_tr, y_te = train_test_split(iris.data, iris.target, test_size=0.3, random_state=42)
gnb = GaussianNB()
gnb.fit(X_tr, y_tr)
print(f'GaussianNB Iris Test Accuracy: {gnb.score(X_te, y_te)*100:.2f}%')

# 2. Multinomial Naive Bayes on Text with Laplace Smoothing (alpha=1.0)
corpus = [
    'free lottery cash prize winner claim',
    'quarterly revenue financial budget forecast',
    'claim your free luxury gift card now',
    'please review the attached report and spreadsheet'
]
labels = [1, 0, 1, 0]  # 1: Spam, 0: Ham

vec = CountVectorizer()
X_text = vec.fit_transform(corpus)
mnb = MultinomialNB(alpha=1.0)  # Laplace smoothing
mnb.fit(X_text, labels)

test_doc = ['free meeting schedule']
pred = mnb.predict(vec.transform(test_doc))[0]
probs = mnb.predict_proba(vec.transform(test_doc))[0]
print(f'\nTest Document: "{test_doc[0]}"')
print(f'Predicted: {"SPAM" if pred==1 else "HAM"} (Probabilities: Ham={probs[0]:.3f}, Spam={probs[1]:.3f})')
```
