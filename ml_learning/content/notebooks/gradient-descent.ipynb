### 📖 Conceptual Foundations

#### The Optimization Problem

In machine learning, learning is formulated as minimizing a scalar objective function $J(\mathbf{w})$.

Gradient Descent is an iterative first-order optimization algorithm. Starting from an initial parameter point $\mathbf{w}_0$, it repeatedly takes steps proportional to the **negative gradient** of the function at the current point:

$$\mathbf{w}_{t+1} = \mathbf{w}_t - \eta \nabla J(\mathbf{w}_t)$$

- $\eta > 0$ is the **learning rate** (step size hyperparameter).
- $\nabla J(\mathbf{w}_t)$ is the gradient vector pointing in the direction of steepest increase.


#### 3D Loss Surface & Gradient Step Control


#### The Three Flavors of Gradient Descent

| Variant | Data Used per Step | Gradient Stability | Speed per Step | Memory Footprint |
|---|---|---|---|---|
| **Batch GD** | All $N$ training samples | Completely deterministic and smooth | Slow on large $N$ | High (must load full dataset) |
| **Stochastic GD (SGD)** | Exactly 1 random sample | Highly noisy and erratic; helps escape shallow saddles | Extremely fast | Tiny ($O(1)$ sample memory) |
| **Mini-Batch GD** | $B$ samples (e.g., 32, 64, 256) | Balanced compromise; smooth enough, exploits GPU parallelism | Optimal for modern hardware | Configurable by batch size $B$ |


#### Learning Rate ($\eta$) Dynamics

- **Too Small ($\eta = 10^{-6}$):** Convergence is painfully slow; gets trapped in local plateaus.
- **Optimal ($\eta = 10^{-2}$):** Smooth, rapid descent toward the global/local minimum.
- **Too Large ($\eta = 10.0$):** Overshoots the minimum, oscillates wildly, and **diverges to infinity** (NaN loss values).


#### Production Standard

Mini-batch Gradient Descent with adaptive momentum (such as Adam or RMSprop) is the universal industry standard for training modern machine learning models and deep neural networks.



```python
# %matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

# Generate synthetic linear dataset: y = 2*x + 1
np.random.seed(42)
X = np.random.rand(100, 1)
y = 2 * X + 1 + np.random.randn(100, 1) * 0.1
X_b = np.c_[np.ones((100, 1)), X]  # add x0 = 1

eta = 0.1
n_iterations = 50
m = 100

# Batch Gradient Descent
w = np.random.randn(2, 1)
loss_history = []

for iteration in range(n_iterations):
    gradients = (2 / m) * X_b.T.dot(X_b.dot(w) - y)
    w = w - eta * gradients
    loss = np.mean((X_b.dot(w) - y)**2)
    loss_history.append(loss)

print(f'Trained Weights after {n_iterations} iterations: Intercept = {w[0][0]:.4f}, Slope = {w[1][0]:.4f}')
print(f'Initial Loss: {loss_history[0]:.4f} -> Final Loss: {loss_history[-1]:.4f}')
```
