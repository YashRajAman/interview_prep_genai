### 📖 Conceptual Foundations

#### The Sigmoid (Logistic) Function

Unlike linear regression which outputs unbounded continuous values ($-\infty$ to $+\infty$), logistic regression passes a linear combination through the **logistic sigmoid function** $\sigma(z)$ to produce a calibrated probability between 0 and 1:

$$z = \mathbf{w}^T \mathbf{x} + b$$
$$\hat{p} = P(y = 1 | \mathbf{x}) = \sigma(z) = \frac{1}{1 + e^{-z}} = \frac{1}{1 + e^{-(\mathbf{w}^T \mathbf{x} + b)}}$$

- When $z \rightarrow +\infty$, $\sigma(z) \rightarrow 1$.
- When $z \rightarrow -\infty$, $\sigma(z) \rightarrow 0$.
- When $z = 0$, $\sigma(z) = 0.5$ (the default decision boundary threshold).


#### Log-Odds and the Logit Transformation

The **odds** of an event with probability $p$ is defined as the ratio of success to failure:
$$\text{Odds} = \frac{p}{1 - p}$$

Taking the natural logarithm of the odds yields the **logit** function:
$$\text{logit}(p) = \ln\left(\frac{p}{1 - p}\right) = \mathbf{w}^T \mathbf{x} + b$$

Thus, Logistic Regression is fundamentally a linear model for the **log-odds** of the positive class!


#### The Cost Function: Binary Cross-Entropy (Log Loss)

We cannot use Mean Squared Error (MSE) for logistic regression because the non-linear sigmoid causes MSE to be non-convex with numerous local minima traps. Instead, we use **Binary Cross-Entropy Loss** derived from Maximum Likelihood Estimation (MLE):

$$J(\mathbf{w}) = -\frac{1}{n} \sum_{i=1}^n \left[ y^{(i)} \ln(\hat{p}^{(i)}) + (1 - y^{(i)}) \ln(1 - \hat{p}^{(i)}) \right]$$

- If ground truth $y = 1$ and model predicts $\hat{p} \rightarrow 1$, penalty $\rightarrow 0$.
- If ground truth $y = 1$ and model predicts $\hat{p} \rightarrow 0$, penalty $-\ln(\hat{p}) \rightarrow +\infty$!
- Binary Cross-Entropy is strictly **convex** with respect to weights $\mathbf{w}$.


#### Multiclass Extensions: OvR vs. Softmax

| Strategy | Architecture | Formulation | Best For |
|---|---|---|---|
| **One-vs-Rest (OvR)** | Trains $K$ separate binary logistic classifiers | $P(y=k \text{ vs not } k)$ | Multilabel classification or large $K$ |
| **Multinomial (Softmax)** | Single unified model producing normalized probability vector | $P(y=k) = \frac{e^{\mathbf{w}_k^T \mathbf{x}}}{\sum_j e^{\mathbf{w}_j^T \mathbf{x}}}$ | Mutually exclusive multiclass problems (e.g. MNIST 0-9) |


#### Decision Threshold Calibration

The default classification threshold is 0.5. However, in asymmetric risk domains (e.g. cancer detection or fraud), the engineer can lower the threshold to 0.15 to maximize **Recall** at the expense of Precision.



```python
# %matplotlib inline
import numpy as np
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression

# Generate a synthetic binary classification dataset
X, y = make_classification(n_samples=100, n_features=2, n_redundant=0, n_informative=2,
                           n_clusters_per_class=1, random_state=42)

# Fit Logistic Regression
clf = LogisticRegression(random_state=42)
clf.fit(X, y)

# Evaluate predicted probabilities for the first 3 samples
probs = clf.predict_proba(X[:3])
preds = clf.predict(X[:3])

print('Learned Coefficients (weights):', clf.coef_[0])
print('Learned Intercept (bias):      ', clf.intercept_[0])
print('\nFirst 3 Samples:')
for i in range(3):
    print(f'Sample {i}: P(Class 0) = {probs[i][0]:.3f}, P(Class 1) = {probs[i][1]:.3f} -> Predicted: {preds[i]}')
```
