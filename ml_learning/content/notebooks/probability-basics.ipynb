### 📖 Conceptual Foundations

#### Probability is Just Measuring Chance

In the real world, things are rarely 100% guaranteed. When a doctor diagnoses a patient or a weather app forecasts tomorrow's weather, they speak in terms of **probability**:

- Probability is always a number between **0.0 (0%)** and **1.0 (100%)**.
- A probability of **0.0** means the event is completely impossible.
- A probability of **1.0** means the event is 100% certain.
- A probability of **0.5 (50%)** means it's an even coin toss!


#### Everyday Probability Examples

| Event | Probability as a Decimal | Probability as a Percentage | Meaning |
| --- | --- | --- | --- |
| Flipping Heads on a fair coin | 0.50 | 50% | Equally likely to be heads or tails. |
| Rolling a 6 on a standard 6-sided die | 0.167 | 16.7% | One chance out of 6. |
| The sun rising tomorrow | 1.00 | 100% | Completely certain. |
| A fish riding a bicycle | 0.00 | 0% | Completely impossible. |


#### The Golden Rule: Correlation Does NOT Mean Causation!

Just because two things rise and fall together (**correlation**), it does NOT mean that one causes the other (**causation**):

- *Real Example:* Ice cream sales and sunburn cases both skyrocket in July. Does eating ice cream cause sunburns? Of course not! Both are caused by a third factor: **hot sunny summer weather**!

Machine learning models find correlations in data, but humans must use common sense to understand real-world cause and effect.



```python
# %matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

# 1. Flip a coin 1,000 times in Python
flips = np.random.choice(['Heads', 'Tails'], size=1000)
heads_percentage = np.mean(flips == 'Heads') * 100
print(f'1,000 Coin Flips -> Heads: {heads_percentage:.1f}% (Expected ~50%)')

# 2. Simulate human heights (Bell Curve: Mean = 170cm, Std = 8cm)
heights = np.random.normal(loc=170, scale=8, size=5000)

print(f'Average Height: {np.mean(heights):.1f} cm')
print(f'Middle Height (Median): {np.median(heights):.1f} cm')

plt.figure(figsize=(6, 4))
plt.hist(heights, bins=30, color='skyblue', edgecolor='black', alpha=0.7)
plt.title('The Classic Bell Curve (Heights)')
plt.xlabel('Height (cm)'); plt.ylabel('Count of People'); plt.grid(True)
plt.show()
```
