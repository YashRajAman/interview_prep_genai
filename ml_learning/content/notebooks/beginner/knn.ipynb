### 📖 Conceptual Foundations

#### Core Algorithm Mechanism

KNN is a **non-parametric, instance-based (lazy)** learning algorithm. It makes zero explicit assumptions about the underlying distribution and performs **no training phase**.

To predict a new query point $\mathbf{x}_q$:
1. Compute the distance between $\mathbf{x}_q$ and all $N$ points in the training dataset.
2. Identify the $K$ closest training points (nearest neighbors).
3. **Classification:** Return the **majority vote** (mode) among the $K$ labels.
4. **Regression:** Return the **mean average** of the $K$ target values.


#### Standard Distance Metrics

- **Euclidean Distance ($L_2$):** Straight-line geometric distance:
$$d(\mathbf{u}, \mathbf{v}) = \sqrt{\sum_{i=1}^d (u_i - v_i)^2}$$

- **Manhattan Distance ($L_1$):** Taxicab grid distance (sum of absolute coordinate differences):
$$d(\mathbf{u}, \mathbf{v}) = \sum_{i=1}^d |u_i - v_i|$$

- **Minkowski Distance ($L_p$):** Generalized distance ($p=1$ is Manhattan, $p=2$ is Euclidean):
$$d(\mathbf{u}, \mathbf{v}) = \left( \sum_{i=1}^d |u_i - v_i|^p \right)^{1/p}$$

- **Cosine Distance:** $1 - \text{CosineSimilarity}(\mathbf{u}, \mathbf{v})$, measuring directional difference independently of vector magnitude.


#### The Impact of Hyperparameter K: Bias-Variance Tradeoff

| Value of K | Model Complexity | Decision Boundary | Risk |
|---|---|---|---|
| **$K = 1$** | Maximum complexity | Extremely jagged, wraps around individual training noise points | **Severe Overfitting** (High Variance) |
| **Optimal $K$ (e.g. 5-15)** | Balanced | Smooth, robust boundary capturing true class regions | **Optimal Generalization** |
| **Large $K$ ($K \rightarrow N$)** | Minimum complexity | Trivial; predicts only the majority class of the entire dataset | **Severe Underfitting** (High Bias) |


#### Computational Bottleneck & Spatial Indexing

Because KNN computes distances to every training sample during inference:
- **Brute Force:** $O(N \cdot d)$ per test query. Slow for large datasets!
- **KD-Tree:** Hierarchical space-partitioning binary tree, achieving $O(\log N)$ search time in low dimensions ($d < 20$).
- **Ball Tree:** Partitions points into nested hyperspheres, outperforming KD-trees in higher-dimensional spaces.


#### Mandatory Requirement: Feature Scaling

KNN is **distance-based**. If feature $x_1$ (Income: $20,000 - $200,000) has a much larger numeric range than feature $x_2$ (Age: 18 - 80), distance calculations will be completely dominated by Income, rendering Age statistically invisible! **Always standardize features (StandardScaler) before KNN!**


#### Key Takeaway

KNN is intuitive and highly non-linear, but scales poorly to large inference workloads ($O(N)$ query time) and collapses in high dimensions ($d > 50$) due to the curse of dimensionality.



```python
# %matplotlib inline
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler

# Load iris dataset
iris = load_iris()
X, y = iris.data, iris.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Standardize features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print('=== Evaluating Accuracy Across Different K Values ===')
for k in [1, 3, 5, 11, 25]:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_scaled, y_train)
    acc = knn.score(X_test_scaled, y_test)
    print(f'K = {k:2d}: Test Accuracy = {acc * 100:.2f}%')
```
