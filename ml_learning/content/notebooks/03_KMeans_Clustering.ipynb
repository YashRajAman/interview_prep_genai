# 🔍 K-Means Clustering & The Elbow Method
Unsupervised clustering and finding optimal clusters $.


```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

X, _ = make_blobs(n_samples=400, centers=4, cluster_std=0.75, random_state=42)

inertias = []
K_range = range(1, 9)
for k in K_range:
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    km.fit(X)
    inertias.append(km.inertia_)

plt.figure(figsize=(8, 4))
plt.plot(K_range, inertias, "o-", color="#7c5cff", lw=2)
plt.title("Elbow Method for Optimal K")
plt.xlabel("Number of Clusters K")
plt.ylabel("Inertia (Sum of Squared Distances)")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Fit optimal K=4
km_final = KMeans(n_clusters=4, n_init=10, random_state=42)
labels = km_final.fit_predict(X)

plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap="viridis", alpha=0.7, s=30)
plt.scatter(km_final.cluster_centers_[:, 0], km_final.cluster_centers_[:, 1], c="red", marker="X", s=180, label="Centroids")
plt.title("K-Means Clustering Result (K=4)")
plt.legend()
plt.tight_layout()
plt.show()
```
