### 📖 Conceptual Foundations

#### Mathematical Formulation

Linear regression models the relationship between a continuous dependent target $y$ and $d$ explanatory features $\mathbf{x} = [x_1, x_2, \dots, x_d]^T$ via a linear hyperplane:

$$\hat{y} = w_0 + w_1 x_1 + w_2 x_2 + \dots + w_d x_d = \mathbf{w}^T \mathbf{x}$$

- $\mathbf{w} \in \mathbb{R}^{d+1}$ is the parameter vector (weights and bias intercept $w_0$).
- $\epsilon = y - \hat{y}$ is the residual error vector representing irreducible stochastic noise.


#### The Cost Function: Mean Squared Error (MSE)

To find the optimal weights $\mathbf{w}^*$, we minimize the sum of squared differences between ground truth $y$ and prediction $\hat{y}$:

$$J(\mathbf{w}) = \frac{1}{2n} \sum_{i=1}^n (y^{(i)} - \hat{y}^{(i)})^2 = \frac{1}{2n} \|\mathbf{y} - X\mathbf{w}\|^2$$

Why square the errors?
1. Squaring penalizes large deviations much more aggressively than small ones.
2. Squaring ensures errors are strictly non-negative.
3. The function is continuous, smooth, and strictly **convex**, ensuring a single unique global minimum with zero local minima traps!

#### Analytical Solution: The Normal Equation

Setting the gradient $\nabla_{\mathbf{w}} J(\mathbf{w}) = \mathbf{0}$ yields the closed-form analytical solution:

$$\mathbf{w}^* = (X^T X)^{-1} X^T \mathbf{y}$$

| Approach | Time Complexity | When to Use |
|---|---|---|
| **Normal Equation** | $O(d^3)$ due to matrix inversion $(X^T X)^{-1}$ | Small to moderate feature count ($d < 10,000$) |
| **Gradient Descent** | $O(k \cdot n \cdot d)$ per $k$ iterations | Large feature count or streaming datasets ($d > 10,000, n > 1,000,000$) |


#### The 5 Classical Gauss-Markov Assumptions of OLS

1. **Linearity in Parameters:** The true relationship between features and target is linear in coefficients.
2. **Strict Exogeneity:** Errors have an expected value of zero given $X$: $\mathbb{E}[\epsilon | X] = 0$.
3. **Homoscedasticity:** Error variance is constant across all feature values: $\text{Var}(\epsilon_i) = \sigma^2$.
4. **No Autocorrelation:** Errors are statistically uncorrelated between samples: $\text{Cov}(\epsilon_i, \epsilon_j) = 0$.
5. **No Perfect Multicollinearity:** The matrix $X^T X$ has full column rank and is invertible.


#### Evaluation Metric: Coefficient of Determination ($R^2$)

$$R^2 = 1 - \frac{SS_{\text{res}}}{SS_{\text{tot}}} = 1 - \frac{\sum (y_i - \hat{y}_i)^2}{\sum (y_i - \bar{y})^2}$$
An $R^2 = 0.85$ means the model explains **85% of the total variance** in the target variable compared to a naive mean baseline.



```python
# %matplotlib inline
import numpy as np
from sklearn.linear_model import LinearRegression

# Generate synthetic linear data: y = 3*x + 4 + noise
np.random.seed(42)
X_raw = 2 * np.random.rand(100, 1)
y = 4 + 3 * X_raw + np.random.randn(100, 1) * 0.5

# 1. Closed-form Normal Equation: w = (X^T * X)^(-1) * X^T * y
# Add bias column of ones
X_b = np.c_[np.ones((100, 1)), X_raw]
w_best = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)

# 2. Scikit-learn LinearRegression
reg = LinearRegression()
reg.fit(X_raw, y)

print(f'From Scratch Weights: Intercept = {w_best[0][0]:.4f}, Slope = {w_best[1][0]:.4f}')
print(f'Scikit-learn Weights: Intercept = {reg.intercept_[0]:.4f}, Slope = {reg.coef_[0][0]:.4f}')

# Calculate R^2 score
y_pred = reg.predict(X_raw)
ss_res = np.sum((y - y_pred)**2)
ss_tot = np.sum((y - np.mean(y))**2)
r2 = 1 - (ss_res / ss_tot)
print(f'R^2 Score: {r2:.4f} (Model explains {r2*100:.2f}% of variance)')
```
