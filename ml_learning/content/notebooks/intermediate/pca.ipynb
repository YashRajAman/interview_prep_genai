### 📖 Conceptual Foundations

#### The Dual Perspectives of PCA

Pioneered by Karl Pearson (1901), PCA transforms $d$ correlated features into $k < d$ orthogonal, uncorrelated **Principal Components** through two mathematically equivalent formulations:
1. **Variance Maximization:** Find orthogonal axes along which the projection of the data has maximum possible variance.
2. **Reconstruction Error Minimization:** Find a lower-dimensional linear subspace that minimizes the squared Euclidean reconstruction distance to the original points.


#### Mathematical Derivation via the Covariance Matrix

1. **Mean Center Data:** Ensure every feature has zero mean: $\bar{\mathbf{x}} = \mathbf{0}$.
2. **Compute Covariance Matrix:** For centered matrix $X \in \mathbb{R}^{n \times d}$:
$$\Sigma = \frac{1}{n-1} X^T X \in \mathbb{R}^{d \times d}$$
3. **Eigen-Decomposition:** Solve for eigenvalues $\lambda_i$ and eigenvectors $\mathbf{v}_i$:
$$\Sigma \mathbf{v}_i = \lambda_i \mathbf{v}_i$$
- The **eigenvectors** $\mathbf{v}_i$ (principal directions) are mutually orthogonal unit vectors: $\mathbf{v}_i^T \mathbf{v}_j = 0$.
- The **eigenvalues** $\lambda_i$ represent the exact variance explained along each principal direction ($PC_1$ has largest $\lambda_1$).


#### Modern Implementation: Singular Value Decomposition (SVD)

Calculating $X^T X$ directly is computationally inefficient and numerically unstable. Modern implementations (including Scikit-learn) perform SVD directly on centered data $X$:

$$X = U \Sigma V^T$$

- The right singular vectors $V$ are the **principal component directions (eigenvectors)**.
- The singular values $s_i$ relate directly to eigenvalues: $\lambda_i = \frac{s_i^2}{n-1}$.
- Transformed data coordinates (scores) are computed by $Z = X V_k$.


#### Explained Variance Ratio & Scree Plot

The proportion of total variance captured by the $j$-th principal component is:
$$\text{Explained Variance Ratio}_j = \frac{\lambda_j}{\sum_{i=1}^d \lambda_i}$$

**Scree Plot Rule of Thumb:** Retain the top $k$ components that explain **85% to 95%** of the cumulative variance, or cut at the visible 'elbow'.


#### Mandatory Requirement: Zero-Centering & Standardization

PCA is entirely driven by variance. If feature $x_1$ has a variance of $10,000$ and $x_2$ has a variance of $1.0$, $PC_1$ will align almost 100% with $x_1$ regardless of whether $x_2$ is informative! **Always standardize features (StandardScaler) before PCA!**


#### PCA vs. t-SNE / UMAP

PCA is a **linear, global** projection method: fast, deterministic, invertible, and excellent for downstream modeling. In contrast, t-SNE and UMAP are **non-linear, localized manifold learning** methods: superb for 2D visualization, but stochastic, non-invertible, and unsuitable for direct feature preprocessing.



```python
# %matplotlib inline
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load 30-dimensional breast cancer dataset
data = load_breast_cancer()
X, y = data.data, data.target

# 1. Standardize features (mandatory for PCA)
X_scaled = StandardScaler().fit_transform(X)

# 2. Fit PCA retaining all 30 components to inspect variance
pca_full = PCA(random_state=42)
pca_full.fit(X_scaled)

cumulative_var = np.cumsum(pca_full.explained_variance_ratio_)
k_90 = np.argmax(cumulative_var >= 0.90) + 1

print(f'Original Dimensions: {X.shape[1]} features')
print(f'PC1 Explains: {pca_full.explained_variance_ratio_[0]*100:.2f}% of total variance')
print(f'PC2 Explains: {pca_full.explained_variance_ratio_[1]*100:.2f}% of total variance')
print(f'Top 2 components together explain {cumulative_var[1]*100:.2f}% variance')
print(f'Components needed to explain >= 90% variance: {k_90} (compressed from 30 to {k_90}!)')

# 3. Project to 2D for visualization
pca_2d = PCA(n_components=2, random_state=42)
X_2d = pca_2d.fit_transform(X_scaled)
print(f'Projected 2D Shape: {X_2d.shape}')
```
