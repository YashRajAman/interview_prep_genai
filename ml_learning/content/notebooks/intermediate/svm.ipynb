### 📖 Conceptual Foundations

#### The Maximum Margin Classifier

For linearly separable binary classification with labels $y \in \{-1, +1\}$, infinite separating hyperplanes exist. SVM finds the unique hyperplane that maximizes the **geometric margin** (distance to the closest training points on either side):

$$\mathbf{w}^T \mathbf{x} + b = 0$$
$$\text{Margin} = \frac{2}{\|\mathbf{w}\|}$$

Maximizing margin $\frac{2}{\|\mathbf{w}\|}$ is mathematically equivalent to minimizing $\frac{1}{2} \|\mathbf{w}\|^2$ subject to the separation constraint:
$$y^{(i)} (\mathbf{w}^T \mathbf{x}^{(i)} + b) \ge 1, \quad \forall i$$


#### Soft Margin SVM & Slack Variables (C)

Real-world data is rarely linearly separable. **Soft Margin SVM** introduces non-negative slack variables $\xi_i \ge 0$ allowing samples to violate the margin:

$$\min_{\mathbf{w}, b, \xi} \frac{1}{2} \|\mathbf{w}\|^2 + C \sum_{i=1}^n \xi_i \quad \text{subject to} \quad y^{(i)}(\mathbf{w}^T \mathbf{x}^{(i)} + b) \ge 1 - \xi_i$$

- **Large $C$ (Hard Margin tendency):** Heavy penalty for margin violations. Narrow margin, low bias, **high variance (risk of overfitting)**.
- **Small $C$ (Soft Margin):** Tolerates more violations. Wide margin, higher bias, **low variance (more robust to noise)**.


#### The Kernel Trick: Non-Linear Mapping

When data is non-separable in original space $\mathbb{R}^d$, we map it into a higher-dimensional feature space $\Phi(\mathbf{x}) \in \mathbb{R}^D$ where it becomes linearly separable.

The **Kernel Trick** computes dot products in the high-dimensional space without ever explicitly calculating coordinates $\Phi(\mathbf{x})$: $K(\mathbf{x}, \mathbf{z}) = \langle \Phi(\mathbf{x}), \Phi(\mathbf{z}) \rangle$!

| Kernel Name | Formula | Hyperparameters | Behavior |
|---|---|---|---|
| **Linear** | $K(\mathbf{x}, \mathbf{z}) = \mathbf{x}^T \mathbf{z}$ | None | Fast; ideal for high-dimensional text ($D > 10,000$) |
| **Polynomial** | $K(\mathbf{x}, \mathbf{z}) = (\gamma \mathbf{x}^T \mathbf{z} + r)^d$ | Degree $d$, coef0 $r$, scale $\gamma$ | Models feature interactions up to degree $d$ |
| **RBF (Gaussian)** | $K(\mathbf{x}, \mathbf{z}) = \exp(-\gamma \|\mathbf{x} - \mathbf{z}\|^2)$ | $\gamma$ (bandwidth) | Infinite-dimensional Hilbert space; powerful non-linear boundary |


#### The Role of Support Vectors

The final decision boundary depends **exclusively on the Support Vectors** (the training points lying directly on or inside the margin). Deleting all other training points has **zero effect** on the optimal hyperplane!


#### SVM Computational Profile

SVM is memory-efficient because only support vectors are retained for prediction. However, training complexity scales between $O(N^2)$ and $O(N^3)$, making standard SVMs impractical for datasets with $N > 100,000$ samples.



```python
# %matplotlib inline
import numpy as np
from sklearn.datasets import make_moons
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler

# Generate non-linear moon dataset
X, y = make_moons(n_samples=100, noise=0.15, random_state=42)
X_scaled = StandardScaler().fit_transform(X)

# 1. Fit Non-Linear RBF Kernel SVM
clf_rbf = SVC(kernel='rbf', C=1.0, gamma='scale', random_state=42)
clf_rbf.fit(X_scaled, y)

# 2. Fit Linear Kernel SVM
clf_lin = SVC(kernel='linear', C=1.0, random_state=42)
clf_lin.fit(X_scaled, y)

print(f'RBF SVM Accuracy:    {clf_rbf.score(X_scaled, y)*100:.1f}%')
print(f'Linear SVM Accuracy: {clf_lin.score(X_scaled, y)*100:.1f}%')
print(f'Number of Support Vectors (RBF): {len(clf_rbf.support_)} out of 100 samples')
print(f'Support Vector indices (first 5): {clf_rbf.support_[:5]}')
```
