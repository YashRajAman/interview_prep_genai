### 📖 Conceptual Foundations

#### The Wisdom of Crowds (Condorcet's Jury Theorem)

If you aggregate the predictions of a group of diverse, independent predictors that each perform slightly better than random guessing ($p > 0.5$), the collective majority vote approaches **100% accuracy** as the number of voters approaches infinity!


#### Bootstrap Aggregation (Bagging)

A single decision tree has low bias but high variance (easily overfits). **Bagging** reduces variance without increasing bias:

1. **Bootstrap Sampling:** Draw $B$ random subsets of size $N$ from the training set **with replacement**. On average, each bootstrap sample contains ~63.2% unique instances, leaving ~36.8% **Out-of-Bag (OOB)**.
2. **Parallel Training:** Fit an independent base estimator (deep decision tree) on each of the $B$ bootstrap samples in parallel.
3. **Aggregation:** Average predictions across all $B$ models:
   - **Classification:** Plurality majority vote: $\hat{y} = \text{mode}(h_1(\mathbf{x}), \dots, h_B(\mathbf{x}))$.
   - **Regression:** Arithmetic mean: $\hat{y} = \frac{1}{B} \sum_{b=1}^B h_b(\mathbf{x})$.


#### Why Random Forests Beat Simple Bagging

If one feature is overwhelmingly strong (e.g. `Income`), simple Bagging trees will all choose `Income` as their root split, producing **highly correlated trees**! The variance of an average of correlated trees does not drop effectively.

**Random Forests (Breiman, 2001) solve this with Feature Subsampling:**
At every single split in every tree, only a random subset of features $m \approx \sqrt{d}$ (classification) or $m \approx d/3$ (regression) is considered. This **decorrelates the trees**, drastically cutting variance!


#### Out-of-Bag (OOB) Error: Free Validation

Because ~36.8% of training samples are never selected in any given bootstrap draw, each tree can be tested on the samples it never saw during training. Averaging these out-of-bag evaluations yields an **unbiased generalization score without requiring a separate validation set**!


#### Industry Takeaway

Random Forest is widely regarded as the ultimate off-the-shelf tabular baseline: it requires virtually zero hyperparameter tuning, requires no feature scaling, handles missing values, and is almost immune to overfitting as the number of trees increases.



```python
# %matplotlib inline
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
import numpy as np

# Load dataset
data = load_breast_cancer()
X, y = data.data, data.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Fit Random Forest with Out-of-Bag scoring enabled
rf = RandomForestClassifier(n_estimators=100, oob_score=True, random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)

print(f'Random Forest Test Accuracy: {rf.score(X_test, y_test)*100:.2f}%')
print(f'Out-Of-Bag (OOB) Score:      {rf.oob_score_*100:.2f}%')

# Print top 5 most important features
indices = np.argsort(rf.feature_importances_)[::-1][:5]
print('\n=== Top 5 Feature Importances ===')
for rank, idx in enumerate(indices, 1):
    print(f'{rank}. {data.feature_names[idx]:25s}: {rf.feature_importances_[idx]:.4f}')
```
