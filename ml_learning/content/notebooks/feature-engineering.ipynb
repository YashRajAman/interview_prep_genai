### 📖 Conceptual Foundations

#### Andrew Ng's Axiom

"Coming up with features is difficult, time-consuming, requires expert knowledge. 'Applied machine learning' is basically **feature engineering**." — Andrew Ng


#### Continuous Feature Scaling Strategies

- **StandardScaler (Z-Score Standardization):** Centered at 0 with unit variance:
$$z = \frac{x - \mu}{\sigma}$$
Ideal for algorithms sensitive to variance scale (Linear/Logistic Regression, SVMs, Neural Networks, PCA).

- **MinMaxScaler (Normalization):** Scales values strictly into a bounded range $[0, 1]$:
$$x_{\text{norm}} = \frac{x - x_{\min}}{x_{\max} - x_{\min}}$$
Preserves zero entries in sparse matrices and bounded image pixel ranges $[0, 255] \rightarrow [0, 1]$.

- **RobustScaler:** Uses median and interquartile range (IQR) to scale data without being distorted by extreme outliers:
$$x_{\text{robust}} = \frac{x - \text{median}}{\text{IQR}}$$


#### Categorical Encoding Techniques

| Encoding Method | Mechanism | Cardinality Limit | Key Strength / Vulnerability |
|---|---|---|---|
| **One-Hot Encoding** | Creates $K$ binary indicator columns | Low ($K < 20$) | Preserves full category independence; causes dimensional explosion if $K$ is large |
| **Ordinal Encoding** | Maps categories to integers ($1, 2, \dots, K$) | Medium | Imposes artificial mathematical linear order; suitable only for naturally ranked categories |
| **Target (Mean) Encoding**| Replaces category with target mean $\mathbb{E}[y|x=c]$ | High ($K > 1,000$) | Very compact 1D representation; severe risk of target leakage without smoothing |
| **Frequency (Count)** | Replaces category with sample frequency count | High | Compact, preserves popularity signal, no target leakage |


#### Handling Severe Class Imbalance

- **Class Weighting:** Adjusts the loss function penalty inversely proportional to class frequency: $w_j = \frac{N}{K \cdot N_j}$. Highly effective with zero computational overhead.
- **SMOTE (Synthetic Minority Over-sampling Technique):** Synthesizes new minority examples along feature-space line segments connecting k-nearest minority neighbors.
- **Undersampling (e.g. NearMiss, Random):** Downsamples majority class to balance ratios, at the risk of discarding potentially valuable information.


#### SMOTE Leakage Warning

**Never apply SMOTE before the train/test split!** Synthesizing samples before splitting causes synthetic points generated from test-set neighbors to leak directly into the training partition, creating unrealistically high test metrics that fail in production.


#### Feature Engineering Philosophy

Domain knowledge drives the best features: ratios (Debt-to-Income), temporal deltas (Days Since Last Purchase), and aggregations (Mean Transaction over 7 Days) routinely beat raw features regardless of algorithm choice.



```python
# %matplotlib inline
import pandas as pd
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

# Create sample dataset with mixed numeric and categorical features
df = pd.DataFrame({
    'age': [25, 45, 35, 50, 23, 38, 62, 29],
    'income': [45000, 120000, 65000, 180000, 32000, 75000, 240000, 52000],
    'city': ['NYC', 'London', 'Tokyo', 'NYC', 'London', 'Tokyo', 'NYC', 'Tokyo'],
    'churn': [0, 1, 0, 1, 0, 0, 1, 0]
})

X = df.drop('churn', axis=1)
y = df['churn']

# Configure ColumnTransformer
preprocessor = ColumnTransformer(transformers=[
    ('num', StandardScaler(), ['age', 'income']),
    ('cat', OneHotEncoder(drop='first', sparse_output=False), ['city'])
])

# Chain with model
model = Pipeline([
    ('prep', preprocessor),
    ('clf', LogisticRegression(class_weight='balanced', random_state=42))
])

model.fit(X, y)
print('Transformed feature names:', model.named_steps['prep'].get_feature_names_out())
print(f'Model Training Score: {model.score(X, y)*100:.1f}%')
```
